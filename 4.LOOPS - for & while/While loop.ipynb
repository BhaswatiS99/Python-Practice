```python
# Q1. Write a while loop that prints numbers from 1 to 10, one per line.
```

```python
count = 1

while count < 11:
    print(count)
    count += 1
```

```python
# Q2. Write a while loop to find the sum of all numbers from 1 to 100. Print the final total.
```

```python
total = 0
i = 1
while i <= 100:
    total += i   # add i to running total
    i += 1
print(total)  # 5050
```

```python
# Q3. Using a while loop, print a countdown from 10 down to 1, then print 'Go!' after the loop ends.
```

```python
count = 11

while count > 0:
    count -= 1
    print(count)  
print("Go!")    
```

```python
# Q4. Write a while loop to print all even numbers between 1 and 20.
```

```python
i = 1

while i <= 20:
    i += 1
    if i % 2 == 0:
        print(i)    
```

```python
# Q5. Using a while loop, calculate the factorial of 6 (i.e., 6! = 6×5×4×3×2×1) and print the result.
```

```python
n = 6
final = 1

while n > 0:
    final *= n
    n -= 1
print(final)
```

```python
sum = 0
n = 1

while n <= 100:
    sum += n
    n += 1
print(sum)
```

```python
# Q7. Print multiplication table of 5 (5×1 to 5×10)
```

```python
i = 1

while i <= 10:
    print(f"5 * {i} = {5*i}")
    i += 1
```

```python
# Q9. Print the first 5 powers of 2 (2¹, 2², 2³...)
```

```python
i = 1

while i <= 5:
    print(f"2^{i} = {2**i}")
    i += 1
```

```python
# Q10. Find the sum of digits of the number 4567
```

```python
num = 4567
digit_sum = 0

while num > 0:
    digit_sum += num % 10
    num = num // 10
print(digit_sum)  
```
