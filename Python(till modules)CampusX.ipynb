Operators
- used to perform operations on variables and values. Python has following operators
  - Arithmatic
  - comparison
  - logical
  - bitwise
  - assignment
  - identity
  - membership

```python
# arithmatic operator
x = 5
y = 3

print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(x % y)           # % : modulus operator - gives remainder
print(x ** y)          # ** : to the power
print(x // y)          # // : convert the result into int
```

```python
# comparison operator

print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
print(x == y)
print(x != y)
```

```python
# logical operators : or, and
   #OR
   # true or true = true
   # true or false = true
   # false or false = false
   # false or true = true
```

```python
x = True
y = False

print(x or x)
print(x or y)
print(y or x)
print(y or y)
```

```python
print(x and x)
print(x and y)
print(y and x)
print(y and y)
```

```python
# bitwise operators - works on binary      

x= 2           # converting values of x and y to binary and then using &
y= 3          

print(x & y)

# 010
# 110
#------
# 010
```

```python
print(x | y)           # using or operator

# 010
# 110
#------
# 110
```

```python
print(x >> 2)
print(y << 3)
```

```python
# assignment operator: =, +=, -=, *=

a = 10                 # assigning a's value as 3
print(a)

a += 3
# a = a + 3
print(a)

a -= 3
# a = a - 3
print(a)

a *= 3
# a = a * 3
print(a)
```

```python
# identity operator : is , is not
a = 3
b = 3

print(a is b)        # true means a and b are in same memory location
```

```python
a= 'hello'
b= 'hello'

print( a is b)
print( a is not b)
```

```python
a= [1,2,3]
b= [1,2,3]

print(a is b)         #result is false because variables having same values doesnot mean they are stored in the same memory location
```

```python
# membership operator: in, not in
# applicable in list, tuple, sets, dictionary
x= "Delhi"

print("D" in x)
print("D" not in x)
```
